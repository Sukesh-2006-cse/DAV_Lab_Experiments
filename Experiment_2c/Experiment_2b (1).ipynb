```python
import pandas as pd
import numpy as np

# Define number of rows for synthetic data
num_rows = 50

# Generate synthetic data for text_df (mimicking 'Google_data (2b.c1).csv')
data_text = {
    'Text_Column': ['text_' + str(i) for i in np.random.randint(1, 100, num_rows)],
    'Value_A': np.random.randint(1, 1000, num_rows),
    'Value_B': np.random.rand(num_rows) * 100
}
text_df = pd.DataFrame(data_text)
# Introduce some NaNs for ffill
text_df.loc[np.random.choice(text_df.index, size=5, replace=False), 'Value_A'] = np.nan

# Generate synthetic data for excel_df (mimicking 'data (2c2).xlsx')
data_excel = {
    'Product': np.random.choice(['A', 'B', 'C', 'D'], num_rows),
    'Quantity': np.random.randint(10, 500, num_rows),
    'Price': np.random.uniform(5.0, 150.0, num_rows)
}
excel_df = pd.DataFrame(data_excel)
# Introduce some NaNs for bfill
excel_df.loc[np.random.choice(excel_df.index, size=5, replace=False), 'Price'] = np.nan

# Generate synthetic data for web_df (mimicking 'countries.csv')
countries = [f'Country_{i}' for i in range(num_rows)]
regions = np.random.choice(['Asia', 'Europe', 'Africa', 'Americas', 'Oceania'], num_rows)
data_web = {
    'Country': countries,
    'Region': regions,
    'Population': np.random.randint(1_000_000, 1_000_000_000, num_rows),
    'GDP_per_Capita': np.random.uniform(1000, 70000, num_rows)
}
web_df = pd.DataFrame(data_web)
# Introduce some NaNs for dropna
web_df.loc[np.random.choice(web_df.index, size=3, replace=False), 'Population'] = np.nan
web_df.loc[np.random.choice(web_df.index, size=2, replace=False), 'GDP_per_Capita'] = np.nan

# Display data
print("text_df Head:\n", text_df.head(), "\n")
print("excel_df Head:\n", excel_df.head(), "\n")
print("web_df Head:\n", web_df.head(), "\n")

# Handle missing values
text_df.ffill(inplace=True)
excel_df.bfill(inplace=True)
web_df.dropna(inplace=True)

print("\n--- After handling missing values ---")
print("text_df Head (after ffill):\n", text_df.head(), "\n")
print("excel_df Head (after bfill):\n", excel_df.head(), "\n")
print("web_df Head (after dropna):\n", web_df.head(), "\n")
```
