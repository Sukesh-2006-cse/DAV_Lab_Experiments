```python
import pandas as pd
import numpy as np
from scipy.stats import skew, kurtosis
# Import Datasets (using synthetic data as original files were not found)
print("Note: Original CSV files not found. Generating synthetic data for demonstration.")
num_rows = 500
np.random.seed(42) # for reproducibility

uci_diabetes_data = {
    "Pregnancies": np.random.randint(0, 18, num_rows),
    "Glucose": np.random.randint(0, 200, num_rows),
    "BloodPressure": np.random.randint(0, 120, num_rows),
    "SkinThickness": np.random.randint(0, 100, num_rows),
    "Insulin": np.random.randint(0, 850, num_rows),
    "BMI": np.round(np.random.uniform(15.0, 70.0, num_rows), 1),
    "DiabetesPedigreeFunction": np.round(np.random.uniform(0.08, 2.5, num_rows), 3),
    "Age": np.random.randint(20, 81, num_rows),
    "Outcome": np.random.randint(0, 2, num_rows)
}
uci_diabetes = pd.DataFrame(uci_diabetes_data)

pima_diabetes_data = {
    "Pregnancies": np.random.randint(0, 18, num_rows),
    "Glucose": np.random.randint(0, 200, num_rows),
    "BloodPressure": np.random.randint(0, 120, num_rows),
    "SkinThickness": np.random.randint(0, 100, num_rows),
    "Insulin": np.random.randint(0, 850, num_rows),
    "BMI": np.round(np.random.uniform(15.0, 70.0, num_rows), 1),
    "DiabetesPedigreeFunction": np.round(np.random.uniform(0.08, 2.5, num_rows), 3),
    "Age": np.random.randint(20, 81, num_rows),
    "Outcome": np.random.randint(0, 2, num_rows)
}
pima_diabetes = pd.DataFrame(pima_diabetes_data)

# Display Dataset Samples
print("UCI Diabetes Dataset Sample:")
print(uci_diabetes.head())
print("\nPima Indians Diabetes Dataset Sample:")
print(pima_diabetes.head())
# Define Relevant Numerical Columns
numerical_columns = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI",
"DiabetesPedigreeFunction", "Age"]
# Univariate Analysis Function
def univariate_analysis(df, columns):
    stats = {}
    for col in columns:
        # Ensure the column exists before processing
        if col not in df.columns:
            print(f"Warning: Column '{col}' not found in DataFrame. Skipping.")
            continue
        stats[col] = {
            "Mean": np.mean(df[col]),
            "Median": np.median(df[col]),
            "Mode": df[col].mode()[0],
            "Variance": np.var(df[col], ddof=1),
            "Standard Deviation": np.std(df[col], ddof=1),
            "Skewness": skew(df[col]),
            "Kurtosis": kurtosis(df[col])
        }
    return pd.DataFrame(stats).T
# Perform Univariate Analysis
uci_stats = univariate_analysis(uci_diabetes, numerical_columns)
pima_stats = univariate_analysis(pima_diabetes, numerical_columns)
# Display Results
print("\nUCI Diabetes Dataset Statistics:")
print(uci_stats)
print("\nPima Indians Diabetes Dataset Statistics:")
print(pima_stats)
```
